```python

import os
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display

```

```python
load_dotenv(override=True)
openai_api_key = os.getenv('OPENAI_API_KEY')
anthropic_api_key = os.getenv('ANTHROPIC_API_KEY')
google_api_key = os.getenv('GOOGLE_API_KEY')


if openai_api_key:
    print(f"OpenAI API Key exists and begins {openai_api_key[:8]}")
else:
    print("OpenAI API Key not set")
    
if anthropic_api_key:
    print(f"Anthropic API Key exists and begins {anthropic_api_key[:7]}")
else:
    print("Anthropic API Key not set (and this is optional)")

if google_api_key:
    print(f"Google API Key exists and begins {google_api_key[:2]}")
else:
    print("Google API Key not set (and this is optional)")
```

```python
openai = OpenAI()


anthropic_url = "https://api.anthropic.com/v1/"
gemini_url = "https://generativelanguage.googleapis.com/v1beta/openai/"

anthropic = OpenAI(api_key=anthropic_api_key, base_url=anthropic_url)
gemini_client = OpenAI(api_key=google_api_key, base_url=gemini_url)
```

```python
gpt_model = "gpt-4.1-mini"
claude_model = "claude-3-5-haiku-latest"
gemini_model = "gemini-2.5-flash-lite"

gpt_system = "You are an assistant named Ben who is very argumentative; you are having a talk with Carla and Damon. \
you disagree with anything in the conversation and you challenge everything,and respond to both people separately, in a snarky way."

claude_system = "You are an AI assistant. You address yourself as Carla. \
You are having a discussion with Ben and Damon.You comment on their responses\
and try to be objective and philosophical. \
If any person is argumentative, \
you try to calm them down and stay objective. If someone is too laid back, you ask them to be more serious. "



gemini_system = """You are an assistant named Damon who is relaxed, cool, and aloof .\
You are conversing with Ben and Carla.
You go off topic with conversations and say colorful flowery language when you direct
your conversation. You avoid arguments."""

```

```python
def call_gpt():
    messages = [{"role":"system", "content":gpt_system}]
    for gpt, claude,gemini in zip(gpt_messages, claude_messages,gemini_messages):
    
        messages.append({"role": "assistant", "content": gpt})
        messages.append({"role": "user", "content": claude})
        messages.append({"role": "user", "content": gemini })
   
    
    response = openai.chat.completions.create(model=gpt_model, messages=messages)
    return response.choices[0].message.content
```

```python
def call_claude():
    messages=[]
    for gpt, claude, gemini in zip(gpt_messages, claude_messages,gemini_messages):
        messages.append({"role": "user", "content": gpt})
        messages.append({"role": "assistant", "content": claude})
        messages.append({"role":"user", "content": gemini})
    messages.append({"role": "user", "content": gpt_messages[-1]})
    response = anthropic.chat.completions.create(model=claude_model, messages=messages, max_tokens=500)
    return response.choices[0].message.content
```

```python



def call_gemini():
    messages=[{"role":"system","content":gemini_system}]
    for gpt, claude, gemini in zip(gpt_messages, claude_messages, gemini_messages):
        messages.append({"role":"assistant","content":gemini})
        messages.append({"role": "user", "content": gpt})
        messages.append({"role": "user", "content": claude})
    messages.append({"role":"user","content": gpt_messages[-1]})
    messages.append({"role":"user", "content":claude_messages[-1]})
    response = gemini_client.chat.completions.create(model=gemini_model, messages=messages)
    return response.choices[0].message.content
```

```python
gpt_messages = ["Hi there"]
claude_messages = ["Good Day!"]
gemini_messages = ["Groovy"]




for i in range(3):
    gpt_next = call_gpt()
    display(Markdown(f"Ben:\n{gpt_next}\n"))
    gpt_messages.append(gpt_next)
    
    claude_next = call_claude()
    display(Markdown(f"Carla:\n{claude_next}\n"))
    claude_messages.append(claude_next)

    gemini_next = call_gemini()
    display(Markdown(f"Damon:\n{gemini_next}\n"))
    gemini_messages.append(gemini_next)
```
